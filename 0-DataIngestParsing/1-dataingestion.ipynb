### Text files (.txt) - The simplest case

```python
os.makedirs("data/text_files", exist_ok=True)
```

```python
sample_text = {
    "data/text_files/python_intro.txt": """Python is a high-level, interpreted programming language known for its simplicity and readability. It supports multiple programming paradigms, including procedural, object-oriented, and functional programming.
    Key features of Python include dynamic typing, automatic memory management, and a large standard library that provides tools suited to many tasks. Python is widely used in web development, data analysis, artificial intelligence, scientific computing, and more. Its extensive community and wealth of third-party libraries make it a versatile choice for developers across various domains.""",

    "data/text_files/machine_learning.txt": """Machine learning is a subset of artificial intelligence that focuses on the development of algorithms and statistical models that enable computers to perform tasks without explicit instructions. It involves training models on data to recognize patterns and make predictions or decisions based on new input.
    There are several types of machine learning, including supervised learning, unsupervised learning, and reinforcement learning. Applications of machine learning span various fields, such as natural language processing, computer vision, recommendation systems, and autonomous vehicles. The success of machine learning relies on the quality and quantity of data, as well as the choice of algorithms and model evaluation techniques.""",
}
for filepath, content in sample_text.items():
    with open(filepath, "w", encoding="utf-8") as f:
        f.write(content)

print("Sample text files created in 'data/text_files' directory.")
```

### Text Loader - Read Single file

```python
from langchain_community.document_loaders import TextLoader

loader = TextLoader("data/text_files/python_intro.txt", encoding="utf-8")
documents = loader.load()
```

```python
type(documents)
```

```python
print(f"Loaded {len(documents)} document(s) from 'python_intro.txt'.")
print(f"Content of the first document:\n{documents[0].page_content}")
print(f"Metadata of the first document:\n{documents[0].metadata}")
```

### Directory Loader - Multiple text files

```python
from langchain_community.document_loaders import DirectoryLoader

dir_loader = DirectoryLoader(
                "data/text_files", 
                glob="**/*.txt",  # used to specify the which file to load from directory based on their names, extension or folder path. Here, we are loading all .txt files from the directory and its subdirectories.
                loader_cls=TextLoader, 
                loader_kwargs={"encoding": "utf-8"},
                show_progress=True
                )
documents = dir_loader.load()

print(f"Loaded {len(documents)} document(s) from 'data/text_files' directory.")
for i, doc in enumerate(documents):
    print(f"\nDocument {i+1}:")
    print(f"Content: {doc.page_content}")
    print(f"Metadata: {doc.metadata}")  
```

### Text Splitting Strategies

```python
from langchain_text_splitters import RecursiveCharacterTextSplitter, CharacterTextSplitter, TokenTextSplitter
print(documents)
```

```python
## Method 1 -> Charcater text splitter
text = documents[0].page_content
text
```

```python
print("Character text splitter")
char_splitter = CharacterTextSplitter(
    separator="\n",  # Split on newline characters
    chunk_size=100,  # Maximum size of each chunk
    chunk_overlap=20,  # Overlap between chunks
    length_function=len  # Function to calculate the length of each chunk
)

char_chunks = char_splitter.split_text(text)
print(f"Number of chunks created: {len(char_chunks)}")
```

```python
char_chunks
```

```python
## Method 2 -> Recursive character text splitter
recursive_splitter = RecursiveCharacterTextSplitter(
                    separators = ["\n\n", "\n", " ", ""],  # List of separators to use for splitting
                    chunk_size = 200,  # Maximum size of each chunk
                    chunk_overlap = 20,  # Overlap between chunks
                    length_function = len  # Function to calculate the length of each chunk
)
recursive_chunks = recursive_splitter.split_text(text)
print(f"Number of chunks created: {len(recursive_chunks)}")
for i, chunk in enumerate(recursive_chunks):
    print(f"\nChunk {i+1}:")
    print(f"length: {len(chunk)}")
    print(chunk)
```

```python
## Method 3 -> Token text splitter
print("\n Token text splitter")
token_splitter = TokenTextSplitter(
    chunk_size=50,  # Maximum size of each chunk in tokens
    chunk_overlap=10,  # Overlap between chunks in tokens
    length_function=len  # Function to calculate the length of each chunk
)

token_chunks = token_splitter.split_text(text)
print(f"Created {len(token_chunks)} token chunks.")
```

```python
for i, chunk in enumerate(token_chunks):
    print(f"\nToken Chunk {i+1}:")
    print(f"length: {len(chunk)}")
    print(chunk)
```
